```python
import numpy as np

print("NaN:", np.isnan(dados.values).sum())
print("Total:", dados.size)
print("Valores válidos:", np.isfinite(dados.values).sum())
```

```python
print("Mínimo:", np.nanmin(dados.values))
print("Máximo:", np.nanmax(dados.values))
print("Média:", np.nanmean(dados.values))
```

```python
recorte_centro = ds_icechunk["0"].isel(
    y=slice(5000, 5256),
    x=slice(5000, 5256),
)

dados_centro = recorte_centro.compute()

print(dados_centro)
print("Shape:", dados_centro.shape)
```

```python
import numpy as np

janelas = {
    "canto_superior": (0, 0),
    "centro": (5000, 5000),
    "canto_inferior": (10000, 10000),
    "meio_esquerda": (5000, 0),
    "meio_direita": (5000, 10000),
}

for nome, (y, x) in janelas.items():
    recorte = ds_icechunk["0"].isel(
        y=slice(y, y + 128),
        x=slice(x, x + 128),
    )

    valores = recorte.compute().values

    validos = np.isfinite(valores).sum()

    print(
        f"{nome:20s} | válidos: {validos:5d}/{valores.size} | min: {np.nanmin(valores) if validos else 'NaN'} | max: {np.nanmax(valores) if validos else 'NaN'}"
    )
```
